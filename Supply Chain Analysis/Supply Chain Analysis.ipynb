```python
df["Inventory Value"] = df["Stock levels"] * df["Price"]
```

```python
df
```

```python
total_inventory_value = df['Inventory Value'].sum()

print("Total Inventory Value:", round(total_inventory_value,2))
```

```python
avg_stock = df['Stock levels'].mean()

print("Average Stock Level:", round(avg_stock,2))
```

```python
inventory_by_product = (
    df.groupby('Product type')['Stock levels']
      .sum()
      .sort_values(ascending=False)
)

print(inventory_by_product)

import matplotlib.pyplot as plt

inventory_by_product.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title("Stock Levels by Product Type")
plt.xlabel("Product Type")
plt.ylabel("Stock Levels")
plt.xticks(rotation=0)
plt.show()
```

```python
inventory_value = (
    df.groupby('Product type')['Inventory Value']
      .sum()
      .sort_values(ascending=False)
)

print(inventory_value)

inventory_value.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title("Inventory Value by Product Type")
plt.ylabel("Inventory Value")
plt.xticks(rotation=0)
plt.show()
```

```python
products_sold = (
    df.groupby('Product type')['Number of products sold']
      .sum()
      .sort_values(ascending=False)
)

print(products_sold)

products_sold.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title("Products Sold by Category")
plt.ylabel("Units Sold")
plt.xticks(rotation=0)
plt.show()
```

```python
df['Inventory Efficiency'] = (
    df['Number of products sold'] /
    df['Stock levels']
)
```
